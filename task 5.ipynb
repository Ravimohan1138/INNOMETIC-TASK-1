```python
def analyze_engagement(likes):

    total_likes = 0

    for like in likes:
        total_likes += like

    print("Total Likes:", total_likes)

    if total_likes >= 1000:
        print("Post Status: Viral Post")

    else:
        print("Post Status: Normal Engagement")


likes = [200, 300, 250, 300]

analyze_engagement(likes)
```

```python
def medicine_stock(stock):

    print("Medicine Stock:", stock)

    if stock < 10:
        print("Status: Low Stock Alert")

    else:
        print("Status: Stock Sufficient")


medicine_stock(6)
```

```python
def rainfall_checker(rainfall_data, required_level):

    total = 0

    for rain in rainfall_data:
        total += rain

    average = total / len(rainfall_data)

    print("Average Rainfall:", average)

    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")

    else:
        print("Rainfall Status: Inadequate Rainfall")


rainfall_data = [70, 75, 68, 76]

rainfall_checker(rainfall_data, 70)
```

```python
def detect_duplicates(usernames):

    unique_users = set(usernames)

    if len(unique_users) != len(usernames):
        print("Duplicate Accounts Found: Yes")

    else:
        print("Duplicate Accounts Found: No")


usernames = ["ravi", "john", "ravi", "alex"]

detect_duplicates(usernames)
```

```python
def check_eligibility(age):

    print("Patient Age:", age)

    if age >= 18:
        print("Eligibility Status: Eligible")

    else:
        print("Eligibility Status: Not Eligible")


check_eligibility(21)
```

```python
def premium_crops(prices):

    premium = []

    for price in prices:

        if price > 2000:
            premium.append(price)

    print("Premium Crops:", premium)


prices = [1500, 2500, 1800, 3200]

premium_crops(prices)
```

```python
def application_health(errors):

    print("Error Count:", errors)

    if errors == 0:
        print("System Status: Healthy")

    elif errors <= 5:
        print("System Status: Minor Issues")

    else:
        print("System Status: Critical Issues")


application_health(7)
```

```python
def attendance_checker(attendance):

    present = 0

    for day in attendance:

        if day == "P":
            present += 1

    percentage = (present / len(attendance)) * 100

    print("Attendance Percentage:", percentage)

    if percentage >= 75:
        print("Exam Eligibility: Eligible")

    else:
        print("Exam Eligibility: Not Eligible")


attendance = ["P", "P", "A", "P", "P"]

attendance_checker(attendance)
```

```python
def electricity_bill(units):

    if units <= 100:
        bill = units * 3

    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)

    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        status = "Low Usage"

    elif bill <= 1500:
        status = "Moderate Usage"

    else:
        status = "High Usage"

    return bill, status


units = 250

bill, status = electricity_bill(units)

print("Electricity Bill:", bill)

print("Usage Status:", status)
```
