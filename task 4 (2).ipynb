```python
def validate_recharge(amount):

    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print("Recharge amount must be at least ₹50")

    elif amount not in valid_plans:
        print("Invalid recharge plan")

    else:
        print("Recharge successful")


while True:

    amount = int(input("Enter recharge amount: "))

    validate_recharge(amount)

    retry = input("Retry? (yes/no): ")

    if retry.lower() != "yes":
        break
```

```python
def check_inventory(products):

    for product, stock in products.items():

        if stock < 15:
            print(product, "-> Reorder Alert")

        else:
            print(product, "-> Stock OK")


products = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 18
}

check_inventory(products)
```

```python
def process_result(marks):

    total = 0

    for mark in marks:
        total += mark

    average = total / len(marks)

    print("Average Marks =", average)

    if average >= 50:
        return "Pass"

    else:
        return "Fail"


marks = [65, 70, 45, 80]

result = process_result(marks)

print("Result =", result)
```

```python
def calculate_fare(distance, peak_hour):

    fare = 50 + (distance * 12)

    if peak_hour.lower() == "yes":
        fare = fare + (fare * 0.25)

    return fare


while True:

    distance = float(input("Enter distance in km: "))

    peak = input("Peak hour? (yes/no): ")

    total_fare = calculate_fare(distance, peak)

    print("Total Fare = ₹", total_fare)

    retry = input("Calculate again? (yes/no): ")

    if retry.lower() != "yes":
        break
```

```python
def check_eligibility(attendance):

    present = 0

    for day in attendance:

        if day == "P":
            present += 1

    percentage = (present / len(attendance)) * 100

    print("Attendance Percentage =", percentage)

    if percentage >= 75:
        return "Eligible"

    else:
        return "Not Eligible"


attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]

result = check_eligibility(attendance)

print("Eligibility Status =", result)
```

```python
def check_password(password):

    special_chars = "@#$"

    has_digit = False
    has_special = False

    if len(password) < 8:
        return "Weak Password"

    for char in password:

        if char.isdigit():
            has_digit = True

        if char in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"

    else:
        return "Weak Password"


password = input("Enter password: ")

result = check_password(password)

print(result)
```
