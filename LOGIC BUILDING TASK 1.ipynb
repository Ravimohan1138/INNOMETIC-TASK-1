```python
import re

def preprocess_text(text):
    if not text or not isinstance(text, str):
        return [], ""

    text = re.sub(r"http\S+|www\S+", "", text)

    text = re.sub(r"\S+@\S+", "", text)

    text = re.sub(r"\d+", "", text)

    text = text.lower()

    text = re.sub(r"(.)\1{2,}", r"\1", text)

    text = re.sub(r"[^a-zA-Z\s]", "", text)

    text = re.sub(r"\s+", " ", text).strip()

    tokens = text.split()

    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence
```

```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for s in sentences:
    tokens, clean = preprocess_text(s)

    print("\nOriginal:", s)
    print("Tokens:", tokens)
    print("Cleaned:", clean)

    results.append((tokens, clean))
```

```python
import numpy as np

for i, (tokens, clean) in enumerate(results):
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_length = np.mean([len(word) for word in tokens]) if tokens else 0

    print(f"\nSentence {i+1}:")
    print("Total Tokens:", total_tokens)
    print("Unique Tokens:", unique_tokens)
    print("Avg Token Length:", round(avg_length, 2))
```

```python
from collections import Counter

all_tokens = []

for tokens, _ in results:
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("\nTop 10 Frequent Words:")
print(counter.most_common(10))

print("\nTop 5 Least Frequent Words:")
print(counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
test_cases = ["", "😂😂😂", "123456"]

for t in test_cases:
    tokens, clean = preprocess_text(t)
    print("\nInput:", t)
    print("Tokens:", tokens)
    print("Clean:", clean)
```
