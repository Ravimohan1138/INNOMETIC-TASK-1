```python
def parking_system(capacity, logs):
    current = 0
    peak = 0

    for log in logs:
        if log == "IN":
            current += 1
        elif log == "OUT" and current > 0:
            current -= 1

        peak = max(peak, current)

    print("Currently Parked Vehicles:", current)

    if current > capacity:
        print("Parking Status: Full 🚨")
    else:
        print("Parking Status: Available")

    print("Peak Usage:", peak)


# Example
parking_system(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
def delivery_time(distance, traffic, weather):
    time = distance * 5   # base: 5 min per km

    if traffic == "High":
        time += 15
    elif traffic == "Medium":
        time += 10

    if weather == "Rainy":
        time += 10
    elif weather == "Storm":
        time += 20

    print("Estimated Delivery Time:", time, "minutes")


# Example
delivery_time(8, "High", "Rainy")
```

```python
def theatre_occupancy(total_seats, booked_seats):
    booked = len(booked_seats)
    percentage = (booked / total_seats) * 100

    print("Occupancy:", int(percentage), "%")

    if percentage == 100:
        print("Show Status: Housefull")
    elif percentage >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")


# Example (150 seats booked out of 200)
theatre_occupancy(200, [1]*150)
```

```python
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)

    print("Average CPU Load:", int(avg), "%")

    if avg < 50:
        print("Server Status: Normal")
    elif avg <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Example
server_load([45, 60, 70, 85, 90])
```

```python
def resource_monitor(resources):
    overused = []

    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Example
resource_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

```python
def event_registration(capacity, registrations):
    if registrations <= capacity:
        print("Confirmed Registrations:", registrations)
        print("Waitlisted Users: 0")
        print("Registration Status: Open")
    else:
        waitlist = registrations - capacity
        print("Confirmed Registrations:", capacity)
        print("Waitlisted Users:", waitlist)
        print("Registration Status: Closed")


# Example
event_registration(100, 105)
```
